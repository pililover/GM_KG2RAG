```python
import json

# Bước 1: Đọc dữ liệu từ file
with open('data/chunked_data.json', 'r', encoding='utf-8') as f:
    chunks = json.load(f)['chunks']

with open('data/grounded_triplets.json', 'r', encoding='utf-8') as f:
    triplets = json.load(f)

valid_ids = {triplet['source_chunk_id'] for triplet in triplets}

filtered_chunks = [chunk for chunk in chunks if chunk['chunk_id'] in valid_ids]

with open('data/filtered_chunks.json', 'w') as f:
    json.dump(filtered_chunks, f, indent=4)

print(f"Đã lọc từ {len(chunks)} chunks → {len(filtered_chunks)} chunks hợp lệ.")
```
